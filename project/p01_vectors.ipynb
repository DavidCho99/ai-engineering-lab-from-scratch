# Task 1: L2 Normalization

The L2 norm of a vector is:

$$
\|x\|_2 = \sqrt{\sum_{i=1}^{n} x_i^2}
$$

Normalization converts a vector into a unit vector:

$$
\hat{x} = \frac{x}{\|x\|_2}
$$

```python
def l2Normalize(x, axis =-1, eps=1e-12):
    """
    Normalize vectors using the L2 norm.
    Parameters
    ----------
    x : np.ndarray
        A 1-D or 2-D array.
    axis : int
        Axis along which to normalize.
    eps : float
        Small constant for numerical stability.
    Returns
    -------
    np.ndarray
        Normalized vectors.
    """
    x = np.asarray(x, dtype=float)
    xNorm =  np.linalg.norm(x, axis = axis, keepdims= True)
    xNorm = np.maximum(xNorm,eps)  # maximum performs element-wise comparison.
    return x/xNorm
```

```python
#Test 1: 1D vectors
x1 = np.array([3,4])
print("Test 1:", l2Normalize(x1))
# Test 2: 2D vectors
x2 = np.array([
    [3.0, 4.0],
    [5.0, 12.0]
])

print("Test 2:\n", l2Normalize(x2))

# Test 3: Zero vector
x3 = np.array([0.0, 0.0])
print("Test 3:", l2Normalize(x3))

```
